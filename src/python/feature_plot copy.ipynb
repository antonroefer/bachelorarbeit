```python
def plot_ft(arr, x, t, title, unit, save=False, fname=None, cmap="jet", dpi=300, min_val=None, max_val=None):
    """
    Erstellt einen einheitlichen imshow Plot mit Colorbar und speichert ihn optional ab.

    Parameter:
    - arr: 2D-Array (z.B. Feature Matrix)
    - x: 1D-Array der x-Achse (Distance)
    - t: 1D-Array der y-Achse (Time)
    - title: Plot Titel
    - unit: Einheit für die Colorbar
    - save: True, wenn Plot gespeichert werden soll
    - fname: Dateiname inkl. Pfad
    - cmap: Colormap (default: "jet")
    - dpi: Auflösung für Export
    """
    fig, ax = plt.subplots(figsize=(15, 6))

    im = ax.imshow(
        arr, cmap=cmap, aspect="auto", extent=[x.min(), x.max(), t.max(), t.min()], vmin=min_val, vmax=max_val
    )

    #ax.set_title(title, fontsize=22)
    ax.set_xlabel("Distance (m)", fontsize=18)
    ax.set_ylabel("Time (ns)", fontsize=18)
    ax.tick_params(labelsize=14)

    # Colorbar
    cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.set_label(unit, fontsize=18)
    cbar.ax.tick_params(labelsize=14)

    plt.tight_layout()

    if save and fname:
        plt.savefig(os.path.join(cwd_path, folder, fname), dpi=dpi)
        print(f"Plot gespeichert unter: {fname}")

    plt.show()
```

```python
from scipy.ndimage import sobel
r, i = rg.calc_instantaneous_phase()
phase = np.angle(r + 1j * i)
phase = np.unwrap(phase, axis=0)
grad_t = sobel(phase, axis=0)
grad_x = sobel(phase, axis=1)
dip = np.rad2deg(np.arctan2(grad_x, grad_t))
dip = rg.calc_mean(dip, x_dis=40, y_dis=40)



plot_ft(
    dip,
    x=x,
    t=t,
    title="Dip",
    unit=None,
    save=save,
    fname="dip.png" if save else None,
    cmap="seismic",
    dpi=300,
    min_val=-np.max(np.abs(dip)),
    max_val= np.max(np.abs(dip))
)
```
